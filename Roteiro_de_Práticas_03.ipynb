```python
# 1
import re

original = 'Oiêe!! Tudo bem??? VX já teve aula com o PRofEssor.... QUal o nome mesmo? Hmmmmm, aCho que é RODRIGO LOVERA!, no @insta dele tem VÁRIOS #tbt!'

texto_limpo = re.sub(r'[^A-Za-zÀ-ÿ\s]', '', original)
# r'[^A-Za-zÀ-ÿ\s]'>>> expressõa regular que define um conjunto de caracteres a serem removidos
#texto_limpo = re.sub(r'[A-Za-zÀ-ÿ\s]', '', original) # ^faz a negação

# '' substitui a expressão regular por uma string vazia

texto_normalizado = texto_limpo.lower()

print(f'Texto Original: {original}')
print(f'Texto Limpo: {texto_limpo}')
print(f'Texto normalizado: {texto_normalizado}')
```

```python
# 2 tokenização
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')

tokens = word_tokenize(texto_normalizado)

print(f'Texto Original: {texto_normalizado}')
print(f'Texto limpo: {texto_limpo}')
print(f'Texto normalizado: {texto_normalizado}')
print(f'Tokens extraidos: {tokens}\n')
```

```python
#3 Remoção de Stopwords

from nltk.corpus import stopwords

nltk.download('stopwords')

stopwords_pt = set(stopwords.words('portuguese'))

print(f'Stopwords: {stopwords_pt}\n')

tokens_sem_stopwords = [palavra for palavra in tokens if palavra.lower() not in stopwords_pt]

print(f'Tokens extraidos: {tokens} + \n quantidade de tokens: {len(tokens)}')
print(f'Tokens extraidos: {tokens_sem_stopwords} + \n quantidade de tokens: {len(tokens_sem_stopwords)}\n')
```

```python
#4Stemming e Lematização

from nltk.stem import RSLPStemmer

nltk.download('rslp')

stemmer = RSLPStemmer()
stemming = [stemmer.stem(palavra) for palavra in tokens_sem_stopwords]
print(f'\n\nTokens extraidos: {tokens_sem_stopwords}')
print(f'\n\nTokens radicais: {stemming}\n\n\n')
```
