```python
import re
import unicodedata
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')

def pipeline_limpeza(texto):
    texto = texto.lower()
    texto = re.sub(r'<.*?>', '', texto)
    texto = re.sub(r'http\S+', '' , texto)
    #texto = unicodedata.normalize('NFD' ,texto)
    texto = re.sub(r'[^\w\s]' , '' , texto)

    tokens = word_tokenize(texto)
    return tokens


def calcular_vocabulario(texto):
    tokens = word_tokenize(texto)
    return len(set(tokens))

texto = '''
<span style="color: #840084; font-family: Verdana, Tahoma, Arial, sans-serif; line-height: 16px;">iirru<u>uw</u>w.!&nbsp;<span style="color: magenta;">♥</span>♥<br>
'Stou akii para falar dessa&nbsp;<span style="color: magenta;">mocinha</span>&nbsp;que;<br>
apesar de conhece-la a&nbsp;<i>pouco</i>&nbsp;tempo,<br>
já é m<u>ui</u>ito&nbsp;<b>especiial</b>&nbsp;pra miim.!<br>
<a href="http://www.orkut.com.br/Interstitial?u=http://perfilperfeito.webnode.com.br/perfilperfeito.webnode.com.br&amp;t=ALNX8GEMGujQA6FOFnCQvdq7jIhWpd8k8RVF6BHPOhTLLepkcou6qJpV89btnA5JBcIcq4CgFTjFWubyjMexdr4Egu0z4qwcLwAAAAAAAAAA" style="color: #9310c7;" target="_blank"><span style="color: magenta;"><u>♥♥♥</u></span></a>, spero que nossa amizade dure&nbsp;<u>ever</u>..<br>
voooc é uma pessoa que é sempre bom ter por perto..<br>
'Fareei o possivel e até mesmo o&nbsp;<i>im</i>possivel;<br>
para que nossa&nbsp;<b>amizade</b>&nbsp;seja&nbsp;<span style="color: magenta;">duradoura</span>,<br>
conte comigo sempre&nbsp;<span style="color: magenta;">floor</span>..<br>
-&nbsp;<b>já faz parte da minha viida</b>..&nbsp;<span style="color: magenta;">♥</span></span>
'''

antes = len(texto.strip())

tokens_limpos = pipeline_limpeza(texto)

depois = len(set(tokens_limpos))

print("Vocabulário antes:", antes)
print("Vocabulario depois:", depois)
print("Resultado:", tokens_limpos)
```
