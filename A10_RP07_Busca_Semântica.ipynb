# Busca Tradicional

#Etapa 1

```python
reviews = [
    "A autonomia da bateria é decepcionante",
    "A câmera possui ótima qualidade",
    "O celular é muito rápido",
    "A tela tem boa resolução",
    "Esse negócio descarréga fácil"
]
busca = "O aparelho descarrega muito rápido"
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

matriz_esparsa = vectorizer.fit_transform(reviews)
query = vectorizer.transform([busca])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

scores = cosine_similarity(query, matriz_esparsa)
best_match_idx = scores.argmax()

print(f"Review idenitficado: {reviews[best_match_idx]}")
print(f"Confiança semântica: {scores[0][best_match_idx]:.4f}")

print("\nRanking de relevância: ")
for i in scores.argsort()[0][::-1]:
  print(f"Score: {scores[0][i]:.4f} | Review: {reviews[i]}")
```

###1 -
Resposta: A busca é muito especifica, fazendo com que os vetores tenham menos similaridade, ou seja, sejam ortogonais. Por isso o unico resultado que não pontuou 0 possui o mesmo adjetivo que da lista de busca.

```python
import spacy

nlp = spacy.load("pt_core_news_md")

consulta = nlp(busca)

print("Resultado da busca semântica:")
for texto in reviews:
  doc_vetorizado = nlp(texto)

  similaridade = consulta.similarity(doc_vetorizado)
  print(f"Score de similaridade: {similaridade:.2f} => '{texto}'")
```

###2-
Resposta: O modelo do spaCy é pré treinado. Durante o treinamento os dados identificam os padrões de ocorrência por meio da proximidade no espaço vetorial e relacionam as palavas.

```python
import numpy as np

doc_a = np.array([3 , 1 , 0])
doc_b = np.array([2 , 0 , 1])

def calcular_cosseno(v1 , v2):
  dot_product = np.dot(v1 , v2)
  norm_a = np.linalg.norm(v1)
  norm_b = np.linalg.norm(v2)

  return dot_product / (norm_a * norm_b)
```

```python
print(calcular_cosseno(doc_a , doc_b))
```

Resposta: Utilizaria busca semântica para identificar embeddings similares e técnicas ANN para conseguir completar a busca, só que já otimizado após uma "limpeza" feita com a busca semântica.
