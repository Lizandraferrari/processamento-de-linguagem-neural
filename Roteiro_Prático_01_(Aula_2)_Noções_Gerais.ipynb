```python
#Exercício 1
def inversa(palavra):
  i = len(palavra) - 1
  invertida = ''
  while i >= 0:
    invertida += palavra[i]
    i -= 1
  return invertida

print(inversa('palavra'))
```

```python
# Exercício 2
def qtd_palavras(sentenca):
  palavras = sentenca.split()
  return len(palavras)

print(qtd_palavras('Escreva uma função que receba uma sentença e retorne o número (inteiro) de palavras que ela contém'))
```

```python
#Exercício 3
def retorna_inicial(sentenca):
  iniciais = []
  for palavra in sentenca.split():
    iniciais.append(palavra[0].upper())
  return iniciais


print(retorna_inicial('bastaram anos na Argentina na alvorada'))
```

```python
#Exercício 4
def alfabeto_usado(sentenca):
  sentenca_tratada = ''.join(sentenca.split())

  lista =[]
  for letra in sentenca_tratada:
    if letra not in lista:
      lista.append(letra)
  return lista

print(alfabeto_usado('teste'))
```
