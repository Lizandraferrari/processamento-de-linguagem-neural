##Desafio 1
Durante o treinamento dos dados, o algoritmo não pode ter contato com as informações estatisticas dos dados de teste. Isso acaba alterando as métricas para que sejam mais positivas, fazendo com que diminua a validade da estimativa. Por isso os dados devem ser separados entre dados de treino e não serem reutilizados.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

reviews_teste = ["Produto veio quebrado", "Ruim", "Muito bom", "Bom para o preço", "Muito bom"]
labels_teste = [1, 0, 1, 0, 1]

reviews = ["Produto ótimo", "Péssimo", "Excelente, adorei", "Quebrado e ruim", "Muito bom"]
labels = [1, 0, 1, 0, 1]

vetorizador = TfidfVectorizer()

X_train = vetorizador.fit_transform(reviews)
X_test = vetorizador.transform(reviews_teste)

modelo = MultinomialNB()
modelo.fit(X_train, labels)

predicoes = modelo.predict(X_test)

print("Acurácia:", accuracy_score(labels_teste, predicoes))
```
